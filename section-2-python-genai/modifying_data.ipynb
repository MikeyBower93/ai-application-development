```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}

df = pd.DataFrame(data)
```

```python
df['City'] = ['New York', 'Los Angeles', 'Chicago']
```

```python
df['Age Group'] = ['Young' if age < 30 else 'Mature' for age in df['Age']]
```

```python
df = df.drop('Age Group', axis=1)
```

```python
df = df.drop(1)
```
